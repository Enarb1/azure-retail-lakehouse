```python
from src.config import ORDERS_RAW_PATH, CUSTOMERS_RAW_PATH, ORDER_ITEMS_RAW_PATH, PAYMENTS_RAW_PATH, PRODUCTS_RAW_PATH
from src.spark_session import get_spark
```

```python
spark = get_spark("BronzeIngestion")
```

```python
orders_raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(str(ORDERS_RAW_PATH))
)

orders_raw_df.printSchema()
```

```python
orders_raw_df.show(5, truncate=False)
```

```python
print("Rows: ", orders_raw_df.count())
print("Columns: ", orders_raw_df.columns)
```
